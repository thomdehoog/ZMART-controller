# An example workflow

An overview scan, then a detailed scan at a few positions. The notebook talks
only to the controller, so it runs on any microscope with a driver. The
[tutorial](tutorial.md) next to it builds a fuller workflow, which finds spots
in the overview and scans only those.

1. Plug in a driver
2. Capture two settings
3. Choose the positions
4. Acquire
5. Disconnect

## 1. Plug in a driver

Here it is the mock driver, a simulated microscope that comes with the
controller, so nothing else is needed. On a real microscope, plug in its
driver instead. The rest of the notebook stays the same, apart from the
setting names in step 2 if that microscope uses other ones.

```python
import zmart_controller

zmart_controller.set_instrument(zmart_controller.mock)
```

## 2. Capture two settings

`get_state()` captures the instrument's settings. Every command answers with
`success` and `content`; the state is the content. Its `changeable` part is what
`set_state()` applies later. We capture one gentle and one detailed setting.

```python
overview = zmart_controller.get_state()["content"]
overview["changeable"]["laser_power"] = 2.0
overview
```

```python
detail = zmart_controller.get_state()["content"]
detail["changeable"]["laser_power"] = 8.0
detail["changeable"]["gain"] = 2.0
detail
```

## 3. Choose the positions

Positions are micrometers from the microscope's origin. `get_xyz()` shows the
current position and how far each axis can travel.

```python
zmart_controller.get_xyz()["content"]
```

```python
positions = [
    {"x": 0.0, "y": 0.0, "z": 0.0},
    {"x": 120.0, "y": 0.0, "z": 0.0},
    {"x": 0.0, "y": 120.0, "z": 0.0},
]
```

## 4. Acquire

Apply a setting, then visit each position and capture. `acquire()` captures
and saves in one call. The content says where the files went. Run it again with
`detail` for the detailed scan.

```python
zmart_controller.set_state(overview)
```

```python
for i, pos in enumerate(positions):
    zmart_controller.set_xyz(pos["x"], pos["y"], pos["z"])
    answer = zmart_controller.acquire(position_label=f"pos_{i:03d}")
    print(answer["success"], answer["content"])
```

## 5. Disconnect

```python
zmart_controller.disconnect()
```
